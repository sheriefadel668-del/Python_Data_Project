```python
# Importing Libraries
import ast
import pandas as pd
from datasets import load_dataset
import matplotlib.pyplot as plt  

# Loading Data
dataset = load_dataset('lukebarousse/data_jobs')
df = dataset['train'].to_pandas()

# Data Cleanup
df['job_posted_date'] = pd.to_datetime(df['job_posted_date'])
df['job_skills'] = df['job_skills'].apply(lambda x: ast.literal_eval(x) if pd.notna(x) else x)
```

```python
df_DS = df[df['job_title_short'] == 'Data Engineer'].copy()
```

```python
df_DS
```

```python
df_DS['job_posted_month_no'] = df_DS['job_posted_date'].dt.month
```

```python
df_explode = df_DS.explode('job_skills')
pivot = df_explode.pivot_table(
    index='job_posted_month_no',
    columns='job_skills',
    aggfunc='size',
    fill_value=0
)
```

```python
pivot 
```

```python
pivot.loc['total_skills'] = pivot.sum()
pivot
```

```python
df_DE_Pivot = pivot[pivot.loc['total_skills'].sort_values(ascending=False).index]
```

```python
df_DE_Pivot = df_DE_Pivot.drop('total_skills')
```
